## 🤔 Quick Recall: Why is this Regression?

In **Challenge 01**, you identified this as a regression problem. Let's verify:

**Question:** What type of ML problem is predicting win ratings?

<details>
<summary>💡 Answer (click to reveal)</summary>

**Regression!** ✅

**Why?**
- The target (`win_rating`) is a **continuous numeric value**
- Examples: -2.5, 0.0, 3.7, 5.2 (not categories!)
- Can be ANY number on a scale (not just whole numbers)

**Not classification because:**
- We're not categorizing players as "Good/Bad"
- We're predicting their exact contribution (could be 3.14159...)

**Decision tree reminder:**
```
Predicting numbers? → Regression
Predicting categories? → Classification
```

**Key insight:** Even though we might round the prediction for display, the underlying value is continuous!

## Simple Modelling


### b) Define X and y


We first need to define our features (X) and our target (y).


Let's start off with a simple example - let `X` be **the number of minutes** a player has played.

You can figure out what `y` should be from the task we are trying to achieve!


````python
# YOUR CODE HERE
import sklearn
from sklearn.linear_model import LinearRegression
X = df[["mp"]]
y = df["win_rating"]
````

### c) Train-Test Split

Before training ANY model, we must split our data into two sets.

***Why?** Like studying for an exam:
- Practice with similar questions (train)
- Test with NEW questions (test)

Memorizing exact test questions = cheating (data leakage)

**⚠️ Golden Rule:** NEVER fit/train on test data!

Use `train_test_split()` to create 4 variables:
- `X_train`, `X_test`: Features (split)
- `y_train`, `y_test`: Target (split)

Have a look at the lecture if you are unsure about the syntax.

Use `random_state=0` so you can compare your final results with your buddy's results. It will make the split the same every time

💡 **Pro tip:** Never set random_state in production, only for learning/debugging!

````python
# YOUR CODE HERE
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 42)
````

````python
print("Données d'entraînement :", X_train.shape)
print("Données de test :", X_test.shape)
````

### d) Scaling the Features

Scale features so they're comparable (e.g., minutes: 0-3000 → pacing: -10 to +10).

**Critical workflow:**
1. `scaler.fit(X_train)` ← Learn stats from TRAINING only
2. `scaler.transform(X_train)` ← Apply to training
3. `scaler.transform(X_test)` ← Apply SAME transformation to test

⚠️ **Never** `fit(X_test)` = data leakage!

````python
#your code here

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaler.fit(X_train)


X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
````

````python
from nbresult import ChallengeResult

result = ChallengeResult(
    'first_split',
    X_train_shape=X_train.shape,
    X_test_shape=X_test.shape,
    y_train_len=len(y_train),
    y_test_len=len(y_test),
    X_train_scaled_shape=X_train_scaled.shape,
    X_test_scaled_shape=X_test_scaled.shape
)

result.write()
print(result.check())
````

### e) Modelling

**Linear Regression** finds the best-fit line through your data.

**How it works:**
1. Analyze training data (minutes vs. win rating)
2. Find the line that best fits the pattern
3. Use that line to predict NEW values

Import and instantiate a `LinearRegression` model called `lin_model`, then fit it:

````python
# YOUR CODE HERE

from sklearn.linear_model import LinearRegression

lin_model = LinearRegression()

lin_model.fit(X_train, y_train)
````

**Done!** 🎉 Model trained in 3 lines.

During `.fit()`: analyzed 3,000 examples, found best-fit line, ready to predict!

💡 Always fit on scaled data

### f) Scoring: How Good is Our Model? 📊

Now we test the model on data it's NEVER seen before (the test set):

Let's evaluate the performance of our model.
Score your fitted model on your **test dataset** and store the result in a variable named `single_score`.

Expected result: ~`0.5400` (or 54% when expressed as percentage)

````python
# YOUR CODE HERE

single_score = lin_model.score(X_test, y_test)

print("R² Score :", single_score)
````

### g) Score Interpretation

**R² = 0.54** means the model explains **54% of variation** in win ratings.

| R² Score | Meaning |
|----------|---------|
| 1.0 | Perfect (suspicious!) |
| 0.8-1.0 | Excellent |
| 0.5-0.8 | Good |
| **0.54** | **Moderate (we're here!)** |
| 0.0 | Useless (just predicting average) |

**Can we do better?** Yes! Add more features in Part 2 👇

````python
from nbresult import ChallengeResult

result = ChallengeResult(
    'first_model',
    single_score=single_score,
    model_type=type(lin_model).__name__
)

result.write()
print(result.check())
````

## ⚠️ What Went Wrong?

We got a prediction of `+5247.83` but that's impossible! (Elite players ≈ +10)

We trained on scaled data, but we predicted on unscaled data!

Whatever you do to training data, you need to do to prediction data!

Scale your `new_df` and give prediction another go

````python
# YOUR CODE HERE


X_train = X_train[feature_names]


scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)


lin_model_2 = LinearRegression()
lin_model_2.fit(X_train_scaled, y_train)


new_df_scaled = scaler.transform(new_df[feature_names])


new_player_prediction = lin_model_2.predict(new_df_scaled)

print("Predicted win rating :", new_player_prediction[0])

````

````python
from nbresult import ChallengeResult

# Get the prediction
new_df_scaled = scaler.transform(new_df)
new_player_prediction = float(lin_model_2.predict(new_df_scaled)[0])

result = ChallengeResult(
    'final_model',
    multiple_score=multiple_score,
    single_score=single_score,
    score_improvement=multiple_score - single_score,
    new_player_prediction=new_player_prediction,
    new_df_shape=new_df.shape
)

result.write()
print(result.check())
````
